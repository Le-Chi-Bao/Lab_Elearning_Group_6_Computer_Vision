# Pipeline xử lý ảnh cơ bản — OpenCV và Pillow
Notebook tổng hợp các module trong `src/`. Logic xử lý nằm trong các file `.py`; notebook chỉ cấu hình, gọi pipeline và xem kết quả.

Các bước: đọc/lưu ảnh, chuyển không gian màu, crop/resize và vẽ hình.

```python
from pathlib import Path
import importlib.util
import sys
from types import SimpleNamespace
from IPython.display import display
from PIL import Image

ROOT = Path.cwd().resolve()
if not (ROOT / 'src').is_dir(): ROOT = ROOT / 'image_processing_lab'
if not (ROOT / 'src').is_dir(): raise FileNotFoundError('Hãy mở notebook từ image_processing_lab hoặc thư mục Lab1.')
SRC = ROOT / 'src'
IMAGE_PATH = ROOT / 'images' / 'sample.jpg'
OUTPUTS = ROOT / 'outputs'
sys.path.insert(0, str(SRC))

def load_module(filename):
    path = SRC / filename
    spec = importlib.util.spec_from_file_location(path.stem, path)
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    return module

read_display = load_module('01_read_display.py')
color_conversion = load_module('02_color_conversion.py')
crop_resize = load_module('03_crop_resize.py')
draw_shapes = load_module('04_draw_shapes.py')
args = SimpleNamespace(image=IMAGE_PATH, no_display=True)
print(f'Ảnh đầu vào: {IMAGE_PATH}')
print(f'Thư mục kết quả: {OUTPUTS}')
```

## Chạy toàn bộ pipeline
Mỗi module được gọi qua `main(args)`, không chép lại thuật toán từ `src/` vào notebook.

```python
read_display.main(args)
color_conversion.main(args)
crop_resize.main(args)
draw_shapes.main(args)
print('Đã chạy xong toàn bộ pipeline.')
```

## Xem ảnh kết quả

```python
output_names = ['01_opencv.png','01_pillow.png','02_grayscale.png','02_hsv_to_bgr.png','02_lab_to_bgr.png','03_crop_center.png','03_resize_300x300.png','03_resize_50_percent.png','04_draw_opencv.png','04_draw_pillow_vietnamese.png']
for name in output_names:
    path = OUTPUTS / name
    if path.exists():
        print(name); display(Image.open(path))
    else: print(f'Chưa có: {name}')
```
