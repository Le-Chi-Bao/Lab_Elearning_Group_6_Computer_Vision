# CHUYỂN ĐỔI ẢNH TỪ RGB SANG GRAYSCLAE

```python
# 1. Chuyển đổi sang ảnh xám
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Hiển thị ảnh xám
cv2.imshow("Anh Grayscale", img_gray)

# Chờ bấm phím để đóng
cv2.waitKey(0)
cv2.destroyAllWindows()

# (Tùy chọn) Lưu ảnh xám lại
cv2.imwrite("anh_xam.png", img_gray)
print("Đã xử lý xong ảnh xám!")
```

# thay đổi kích thước theo tỉ lệ

```python
# 4.2 - Resize theo tỉ lệ
scale = 0.3 # ( 0.3 cứ việc thay đổi )


resized_ratio = cv2.resize(
    img,
    None,
    fx=scale,
    fy=scale
)

cv2.imshow("Resize theo ti le", resized_ratio)
cv2.imwrite("anh_resize_50.jpg", resized_ratio)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# CÂU 5

```python
import cv2

# Tạo bản sao từ ảnh gốc
img_line = img.copy()

# Vẽ đường thẳng: từ (50, 50) đến (400, 50), màu đỏ (0,0,255), độ dày 5
cv2.line(img_line, (50, 50), (400, 50), (0, 0, 255), 5)

# Hiển thị
cv2.imshow("1. Duong thang", img_line)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 2. Vẽ hình chữ nhật: từ góc trên-trái (100, 100) đến góc dưới-phải (300, 300), màu xanh lá (0,255,0), độ dày 3
# Tạo bản sao mới từ ảnh gốc
img_rect = img.copy()

# Vẽ hình chữ nhật: từ góc trên-trái (100, 100) đến góc dưới-phải (300, 300), màu xanh lá (0,255,0), độ dày 3
cv2.rectangle(img_rect, (100, 100), (300, 300), (0, 255, 0), 3)

# Hiển thị
cv2.imshow("2. Hinh chu nhat", img_rect)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 3. Vẽ hình tròn: tâm tại (500, 200), bán kính 80, màu xanh dương (255,0,0), tô đặc (-1)
# Tạo bản sao mới từ ảnh gốc
img_circle = img.copy()

# Vẽ hình tròn: tâm tại (500, 200), bán kính 80, màu xanh dương (255,0,0), tô đặc (-1)
cv2.circle(img_circle, (500, 200), 80, (255, 0, 0), -1)

# Hiển thị
cv2.imshow("3. Hinh tron", img_circle)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Tạo bản sao mới từ ảnh gốc
img_text = img.copy()

# Thêm văn bản
cv2.putText(
    img_text,                      # Ảnh
    "WELCOME TO 4 BEE HOME ",                # Nội dung chữ
    (150, 100),                    # Tọa độ bắt đầu viết
    cv2.FONT_HERSHEY_SIMPLEX,      # Kiểu font chữ
    1.5,                           # Tỷ lệ kích thước chữ
    (0, 0, 0),                     # Màu chữ (Đen)
    3,                             # Độ dày nét chữ
    cv2.LINE_AA                    # Kiểu đường (giúp chữ mượt hơn)
)

# Hiển thị
cv2.imshow("4. Van ban", img_text)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Lưu lại bức ảnh có chữ này
cv2.imwrite("anh_co_chu.png", img_text)
print("Đã lưu ảnh có chữ thành công!")
```
